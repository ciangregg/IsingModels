```python
import numpy as np
np.set_printoptions(legacy='1.25')

import math
import matplotlib.pyplot as plt
from numba import njit
```

```python
def energy(s, J, h):
    #2D Ising energy with periodic boundary conditions.
    
    E = 0.0
    N = s.shape[0]

    for i in range(N):
        for j in range(N):
            E -= J * s[i, j] * (
                s[(i + 1) % N, j] +
                s[i, (j + 1) % N]
            )
            E -= h * s[i, j]

    return E

@njit
def sweep_once(s, N, J, h, T, E):
    for _ in range(N * N):
        x = np.random.randint(N)
        y = np.random.randint(N)

        nn = (
            s[(x + 1) % N, y] +
            s[(x - 1) % N, y] +
            s[x, (y + 1) % N] +
            s[x, (y - 1) % N]
        )

        dE = 2 * s[x, y] * (J * nn + h)

        if dE <= 0 or np.random.rand() < math.exp(-dE / T):
            s[x, y] *= -1
            E += dE

    return E
```

```python
def sim_annealing_ising(N,J,h,Tstart,Tend,alpha,sweeps_per_T):
    
    s = np.random.choice(np.array([-1, 1]), size=(N, N))

    E = energy(s, J, h)
    
    temps=[]
    energies=[]
    mags =[]
    
    T=Tstart
    
    while T > Tend:
        
        for _ in range(sweeps_per_T):
            E = sweep_once(s, N, J, h, T, E)
            
        temps.append(T)
        energies.append(E)
        mags.append(s.sum())
        
        T *= alpha
    
    return np.array(temps), np.array(energies), np.array(mags) , s.copy()
```

```python
temps, energies, mags, final_state = sim_annealing_ising(
    N=50,
    J=5,
    h=0,
    Tstart=15,
    Tend=5,
    alpha=0.99,
    sweeps_per_T=1000
)

print("Temps")
print(temps)
print("           ")
print("Energies")
print(energies)
print("           ")

print("magentisation")
print(mags)
print("           ")

print("the final state")
print(final_state)
```

```python
# Final energy vs Sweeps per Temp

#sweeps_per =  np.linspace(100 , 1000 , 5)
sweeps_per =[100,500,750,1000,2000]
ground_E_mean= []
ground_E_std=[]

for per in sweeps_per:
    for _ in range(100):
        ground_E = []
        
        temps, energies, mags, final_state = sim_annealing_ising(
        N=50,
        J=5,
        h=0,
        Tstart=20,
        Tend=5,
        alpha=0.9,
        sweeps_per_T=per) 
        
        ground_E.append(energies[-1])
        
    ground_E_mean.append(np.mean(ground_E))
    ground_E_std.append(np.std(ground_E))


print(f"sweeps {sweeps_per}")
print("gsE")
print(ground_E_mean)
print(ground_E_std)
```

```python
sweeps_per = [100,500, 1000,1500,2000]

how_many= 10
n = 50
j = 5

E_true = -2 * j * n**2
print(E_true)

success = []

for per in sweeps_per:

    success_per = 0

    for _ in range(how_many):

        temps, energies, mags, final_state = sim_annealing_ising(
            N=n,
            J=j,
            h=0,
            Tstart=20,
            Tend=0.1,
            alpha=0.9,
            sweeps_per_T=per
        )

        final_E = energies[-1]

        if final_E == E_true:
            success_per += 1

    success.append(success_per / how_many)

print("Sweeps:", sweeps_per)
print("Success probability:", success)
```

```python
plt.figure(figsize=(8,5))


plt.scatter(sweeps_per,success)
plt.xlabel(r"sweeps per temp")
plt.ylabel(r"success of reaching  $E_\text{ground}$")
plt.title("")
plt.grid(True)
#plt.legend()

#plt.savefig("/home/cian/Documents/college/SURE/git/QUBO/Ising/figures/.pdf")
plt.show()
```
